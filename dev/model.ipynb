# Vocabulary

```python
from pathlib import Path

filepath_vocab = Path("../data/vocab.txt")
filepath_vocab.exists()
vocab = [line.strip() for line in filepath_vocab.open().read().splitlines()]
len(vocab)
```

# Hyperparams

```python
from hlm12nli import Hlm12NliHyperparams, Hlm12NliTokeniserHyperparams

hyperparams = Hlm12NliHyperparams(tokeniser=Hlm12NliTokeniserHyperparams(vocab=vocab))
len(hyperparams.tokeniser.vocab)
```

# Tokenizer

```python
from hlm12nli import Hlm12NliTokeniser

tokeniser = Hlm12NliTokeniser(hyperparams=hyperparams)
tokeniser.detokenise(tokeniser([
    "hi",
    "Good bye and thanks for all the fish.",
    "Farewell, and thanks for all the fish",
]))
```

# Dataset

```python
from hlm12nli import NliDataset, load_nli_dataset

ds_train =  load_nli_dataset(split="train", tokeniser=tokeniser)

print(len(ds_train))
```

```python
anchor, positive, negative = ds_train[:5]
tokeniser.detokenise(anchor), tokeniser.detokenise(positive), tokeniser.detokenise(negative)
```

# Encoder

```python
import torch
from hlm12nli import Hlm12NliEncoder

encoder = Hlm12NliEncoder(hyperparams=hyperparams)
print(encoder)
with torch.no_grad():
    print(encoder(tokeniser([
        "Hi",
        "Good bye and thanks for all the fish.",
        "Farewell, and thanks for all the fish",
    ])))
```
